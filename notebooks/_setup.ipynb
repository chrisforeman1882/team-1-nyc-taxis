## Project Path Setup

This notebook is `%run` by every ETL notebook to ensure `src/` is importable.
It works in both contexts:
- **Manual runs** from `team-1-nyc-taxis/notebooks/`
- **Bundle-deployed runs** from `~/.bundle/.../files/notebooks/`

```python
import sys, os

# Get the workspace path of the *calling* notebook (the one that did %run ./_setup).
# Go up one level from notebooks/ to reach the project root where src/ lives.
_nb_path = dbutils.notebook.entry_point.getDbutils().notebook().getContext().notebookPath().get()
_project_root = "/Workspace" + "/".join(_nb_path.split("/")[:-2])

if _project_root not in sys.path:
    sys.path.insert(0, _project_root)

print(f"Project root: {_project_root}")
```
